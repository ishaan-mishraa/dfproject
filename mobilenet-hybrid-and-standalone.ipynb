```python
#!/usr/bin/env python
import os
import sys
import time
import copy
import random
import numpy as np
from PIL import Image
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms
import timm
from tqdm.auto import tqdm
from sklearn.metrics import (
    accuracy_score, confusion_matrix, precision_score, recall_score,
    f1_score, roc_auc_score, classification_report, roc_curve
)
from scipy.interpolate import interp1d
from scipy.optimize import brentq
import matplotlib.pyplot as plt
import logging
import torch.nn.functional as F
import albumentations as A
from albumentations.pytorch import ToTensorV2

##############################################
#            Setup & Logging                 #
##############################################
seed = 42
torch.manual_seed(seed)
np.random.seed(seed)
random.seed(seed)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s [%(levelname)s] %(message)s",
    handlers=[logging.StreamHandler(sys.stdout)]
)
logging.info(f"Using device: {device}")
if torch.cuda.is_available():
    logging.info(f"GPU: {torch.cuda.get_device_name(0)}")

##############################################
#         Advanced Data Augmentation         #
##############################################
def strong_aug(p=0.5):
    return A.Compose([
        A.RandomRotate90(p=0.2),
        A.Transpose(p=0.2),
        A.HorizontalFlip(p=0.5),
        A.VerticalFlip(p=0.5),
        A.OneOf([A.GaussNoise()], p=0.2),
        A.ShiftScaleRotate(p=0.2),
        A.OneOf([
            A.CLAHE(clip_limit=2),
            A.Sharpen(),
            A.Emboss(),
            A.RandomBrightnessContrast(),
        ], p=0.2),
        A.HueSaturationValue(p=0.2),
        A.RandomCrop(width=200, height=200, p=0.3),
    ], p=p)

def augment(aug, image):
    return aug(image=image)["image"]

class Aug(object):
    def __call__(self, img):
        aug = strong_aug(p=0.9)
        img_np = np.array(img)
        augmented = augment(aug, img_np)
        return Image.fromarray(augmented)

def normalize_data():
    mean = [0.485, 0.456, 0.406]
    std = [0.229, 0.224, 0.225]
    return {
        "train": transforms.Compose([
            Aug(),
            transforms.Resize((224, 224)),
            transforms.RandomHorizontalFlip(),
            transforms.ToTensor(),
            transforms.Normalize(mean, std),
            transforms.RandomErasing(p=0.2)
        ]),
        "valid": transforms.Compose([
            transforms.Resize((224, 224)),
            transforms.ToTensor(),
            transforms.Normalize(mean, std)
        ]),
        "test": transforms.Compose([
            transforms.Resize((224, 224)),
            transforms.ToTensor(),
            transforms.Normalize(mean, std)
        ])
    }

##############################################
#         Dataset Loading & Splitting        #
##############################################
class ExtractedFramesDataset(Dataset):
    def __init__(self, root_dir, transform=None, split="train", random_seed=42):
        """
        Expects root_dir to have the following structure:
        
        root_dir/
            celebdf/
                fake/    <-- label 1
                real/    <-- label 0
            dfdc/
                fake/
                real/
            ffpp/
                fake/
                real/
            realfake/
                fake/
                real/
        """
        self.samples = []
        self.transform = transform

        all_samples = []
        for subfolder in os.listdir(root_dir):
            subfolder_path = os.path.join(root_dir, subfolder)
            if not os.path.isdir(subfolder_path):
                continue
            for label_folder in os.listdir(subfolder_path):
                label_path = os.path.join(subfolder_path, label_folder)
                if not os.path.isdir(label_path):
                    continue
                label = 1 if label_folder.lower() == "fake" else 0
                for file in os.listdir(label_path):
                    if file.lower().endswith(('.png', '.jpg', '.jpeg')):
                        img_path = os.path.join(label_path, file)
                        all_samples.append((img_path, label))
        random.seed(random_seed)
        random.shuffle(all_samples)
        self.samples = all_samples
        print(f"Loaded {len(self.samples)} samples for the '{split}' split.")

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        img_path, label = self.samples[idx]
        image = Image.open(img_path).convert("RGB")
        if self.transform:
            image = self.transform(image)
        return image, label

##############################################
#     Helper to unify feature shapes         #
##############################################
def reduce_features(feat):
    if feat.ndim == 4:
        feat = F.adaptive_avg_pool2d(feat, 1).view(feat.size(0), -1)
    elif feat.ndim == 3:
        feat = feat.mean(dim=1)
    return feat

##############################################
#           MobileNet Model Code             #
##############################################
class MobileNetClassifier(nn.Module):
    def __init__(self, num_classes=1):
        super(MobileNetClassifier, self).__init__()
        self.mobilenet = timm.create_model('mobilenetv3_large_100', pretrained=True)
        self.mobilenet.classifier = nn.Identity()
        self.classifier = nn.Sequential(
            nn.Linear(1280, 512),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(512, 128),
            nn.ReLU(),
            nn.Dropout(0.2),
            nn.Linear(128, num_classes)
        )

    def forward(self, x):
        features = self.mobilenet(x)
        out = self.classifier(features)
        return out

##############################################
#          Label smoothing BCE loss          #
##############################################
class SmoothBCEWithLogitsLoss(nn.Module):
    def __init__(self, smoothing=0.1, pos_weight=None):
        super(SmoothBCEWithLogitsLoss, self).__init__()
        self.smoothing = smoothing
        self.pos_weight = pos_weight

    def forward(self, logits, target):
        target = target * (1 - self.smoothing) + 0.5 * self.smoothing
        return nn.functional.binary_cross_entropy_with_logits(logits, target, pos_weight=self.pos_weight)

##############################################
#          Mixup Data Functions              #
##############################################
def mixup_data(x, y, alpha=0.4):
    if alpha > 0:
        lam = np.random.beta(alpha, alpha)
    else:
        lam = 1
    batch_size = x.size()[0]
    index = torch.randperm(batch_size).to(device)
    mixed_x = lam * x + (1 - lam) * x[index, :]
    y_a, y_b = y, y[index]
    return mixed_x, y_a, y_b, lam

def mixup_criterion(criterion, pred, y_a, y_b, lam):
    return lam * criterion(pred, y_a) + (1 - lam) * criterion(pred, y_b)

##############################################
#       Training & Validation Loops          #
##############################################
def optimize_threshold(probs, true_labels):
    best_thresh = 0.5
    best_acc = 0.0
    for t in np.arange(0.0, 1.0, 0.01):
        preds = (probs >= t).astype(int)
        acc = accuracy_score(true_labels, preds)
        if acc > best_acc:
            best_acc = acc
            best_thresh = t
    return best_thresh, best_acc

##############################################
#                Main Training               #
##############################################
def main():
    data_transforms = normalize_data()
    train_dir = "/kaggle/input/deepfake-dataset/deepfake_dataset/train"
    val_dir = "/kaggle/input/deepfake-dataset/deepfake_dataset/validation"
    test_dir = "/kaggle/input/deepfake-dataset/deepfake_dataset/test"
    
    for path in [train_dir, val_dir, test_dir]:
        if not os.path.exists(path):
            raise FileNotFoundError(f"Dataset directory not found: {path}")
    
    train_dataset = ExtractedFramesDataset(
        train_dir,
        transform=data_transforms["train"],
        split="train"
    )
    val_dataset = ExtractedFramesDataset(
        val_dir,
        transform=data_transforms["valid"],
        split="val"
    )
    test_dataset = ExtractedFramesDataset(
        test_dir,
        transform=data_transforms["test"],
        split="test"
    )

    train_loader = DataLoader(train_dataset, batch_size=16, shuffle=True, num_workers=2)
    val_loader = DataLoader(val_dataset, batch_size=16, shuffle=False, num_workers=2)
    test_loader = DataLoader(test_dataset, batch_size=16, shuffle=False, num_workers=2)

    model = MobileNetClassifier(num_classes=1).to(device)
    logging.info("MobileNet Classifier Model created")

    optimizer = optim.AdamW(model.parameters(), lr=1e-4, weight_decay=1e-4)
    num_epochs = 15
    scheduler = optim.lr_scheduler.LambdaLR(optimizer, lr_lambda=lambda epoch: 0.95 ** epoch)

    pos_weight_value = 10667 / 9333
    criterion = SmoothBCEWithLogitsLoss(
        smoothing=0.1,
        pos_weight=torch.tensor([pos_weight_value]).to(device)
    )

    best_val_loss = float('inf')
    patience = 5
    trigger_times = 0

    train_history = {"loss": [], "acc": []}
    val_history = {"loss": [], "acc": []}

    for epoch in range(num_epochs):
        model.train()
        running_loss = 0.0
        train_correct = 0
        train_total = 0
        train_bar = tqdm(train_loader, desc=f"Epoch {epoch+1}/{num_epochs} [Train]", leave=False)
        for images, labels in train_bar:
            images = images.to(device)
            labels = labels.float().unsqueeze(1).to(device)
            optimizer.zero_grad()

            if random.random() < 0.5:
                images, targets_a, targets_b, lam = mixup_data(images, labels)
                outputs = model(images)
                loss = mixup_criterion(criterion, outputs, targets_a, targets_b, lam)
            else:
                outputs = model(images)
                loss = criterion(outputs, labels)

            loss.backward()
            optimizer.step()

            running_loss += loss.item() * images.size(0)
            preds = (torch.sigmoid(outputs) > 0.5).int()
            train_correct += (preds.cpu() == labels.cpu().int()).sum().item()
            train_total += labels.size(0)
            train_bar.set_postfix(loss=f"{loss.item():.4f}", acc=f"{(train_correct/train_total):.4f}")

        epoch_loss = running_loss / train_total
        epoch_acc = train_correct / train_total
        train_history["loss"].append(epoch_loss)
        train_history["acc"].append(epoch_acc)

        # Validation
        model.eval()
        val_loss = 0.0
        val_correct = 0
        val_total = 0
        val_probs = []
        val_true = []
        val_bar = tqdm(val_loader, desc=f"Epoch {epoch+1}/{num_epochs} [Val]", leave=False)
        with torch.no_grad():
            for images, labels in val_bar:
                images = images.to(device)
                labels = labels.float().unsqueeze(1).to(device)
                outputs = model(images)
                loss = criterion(outputs, labels)

                val_loss += loss.item() * images.size(0)
                probs = torch.sigmoid(outputs)
                val_probs.extend(probs.cpu().numpy().flatten())
                preds = (probs > 0.5).int()
                val_correct += (preds.cpu() == labels.cpu().int()).sum().item()
                val_total += labels.size(0)
                val_true.extend(labels.cpu().numpy().flatten())

        epoch_val_loss = val_loss / val_total
        epoch_val_acc = val_correct / val_total
        val_history["loss"].append(epoch_val_loss)
        val_history["acc"].append(epoch_val_acc)
        scheduler.step()

        best_thresh, _ = optimize_threshold(np.array(val_probs), np.array(val_true))

        print(f"\nEpoch [{epoch+1}/{num_epochs}]")
        print(f"Train Loss: {epoch_loss:.4f}, Train Acc: {epoch_acc:.4f}")
        print(f"Val   Loss: {epoch_val_loss:.4f}, Val Acc: {epoch_val_acc:.4f}, Best Thresh: {best_thresh:.2f}")

        if epoch_val_loss < best_val_loss:
            best_val_loss = epoch_val_loss
            best_model_wts = copy.deepcopy(model.state_dict())
            trigger_times = 0
            torch.save(model.state_dict(), "best_mobilenet.pth")
            logging.info("Best model saved")
        else:
            trigger_times += 1
            if trigger_times >= patience:
                logging.info("Early stopping!")
                break

        checkpoint = {
            "epoch": epoch,
            "model_state_dict": model.state_dict(),
            "optimizer_state_dict": optimizer.state_dict(),
            "scheduler_state_dict": scheduler.state_dict(),
            "best_val_loss": best_val_loss,
            "train_history": train_history,
            "val_history": val_history,
        }
        torch.save(checkpoint, "checkpoint.pth")
        logging.info(f"Checkpoint saved at epoch {epoch+1}")

    def show_plot(metric_history, metric_name):
        plt.figure(figsize=(10, 5))
        plt.plot(metric_history, label=f"{metric_name}")
        plt.title(f"{metric_name} Trend")
        plt.xlabel("Epoch")
        plt.ylabel(metric_name)
        plt.legend()
        plt.show()

    show_plot(train_history["loss"], "Train Loss")
    show_plot(train_history["acc"], "Train Accuracy")
    show_plot(val_history["loss"], "Validation Loss")
    show_plot(val_history["acc"], "Validation Accuracy")

    model.load_state_dict(torch.load("best_mobilenet.pth"))
    model.eval()
    test_probs = []
    all_preds = []
    all_labels = []
    test_loss = 0.0
    test_correct = 0
    test_total = 0

    with torch.no_grad():
        for images, labels in tqdm(test_loader, desc="Testing", leave=False):
            images = images.to(device)
            labels = labels.float().unsqueeze(1).to(device)
            outputs = model(images)
            loss = criterion(outputs, labels)
            test_loss += loss.item() * images.size(0)

            probs = torch.sigmoid(outputs)
            test_probs.extend(probs.cpu().numpy().flatten())
            preds = (probs > best_thresh).int()
            all_preds.extend(preds.cpu().numpy().flatten())
            all_labels.extend(labels.cpu().numpy().flatten())
            test_correct += (preds.cpu() == labels.cpu().int()).sum().item()
            test_total += labels.size(0)

    test_loss /= len(test_dataset)
    accuracy_metric = accuracy_score(np.array(all_labels).astype(int), np.array(all_preds))
    precision_metric = precision_score(np.array(all_labels).astype(int), np.array(all_preds))
    recall_metric = recall_score(np.array(all_labels).astype(int), np.array(all_preds))
    f1_metric = f1_score(np.array(all_labels).astype(int), np.array(all_preds))
    conf_matrix = confusion_matrix(np.array(all_labels).astype(int), np.array(all_preds))
    try:
        auc_metric = roc_auc_score(np.array(all_labels).astype(int), test_probs)
    except:
        auc_metric = "N/A"

    fpr, tpr, thresholds = roc_curve(np.array(all_labels).astype(int), test_probs)
    try:
        eer_metric = brentq(lambda x: 1. - x - interp1d(fpr, tpr)(x), 0., 1.)
    except:
        eer_metric = 1.0

    print("\n=== Test Evaluation Metrics ===")
    print(f"Test Loss: {test_loss:.4f}")
    print(f"Test Accuracy: {accuracy_metric:.4f}")
    print(f"Precision: {precision_metric:.4f}")
    print(f"Recall: {recall_metric:.4f}")
    print(f"F1 Score: {f1_metric:.4f}")
    print(f"ROC-AUC: {auc_metric}")
    print(f"EER: {eer_metric:.4f}")
    print("Confusion Matrix:")
    print(conf_matrix)
    print("Classification Report:")
    print(classification_report(np.array(all_labels).astype(int), np.array(all_preds), digits=4))

if __name__ == "__main__":
    main()
```

```python
#!/usr/bin/env python
import os
import sys
import time
import copy
import random
import numpy as np
from PIL import Image
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms
import timm
from tqdm.auto import tqdm
from sklearn.metrics import (
    accuracy_score, confusion_matrix, precision_score, recall_score,
    f1_score, roc_auc_score, classification_report, roc_curve
)
from scipy.interpolate import interp1d
from scipy.optimize import brentq
import matplotlib.pyplot as plt
import logging
import torch.nn.functional as F
import albumentations as A
from albumentations.pytorch import ToTensorV2

##############################################
#            Setup & Logging                 #
##############################################
seed = 42
torch.manual_seed(seed)
np.random.seed(seed)
random.seed(seed)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s [%(levelname)s] %(message)s",
    handlers=[logging.StreamHandler(sys.stdout)]
)
logging.info(f"Using device: {device}")
if torch.cuda.is_available():
    logging.info(f"GPU: {torch.cuda.get_device_name(0)}")

##############################################
#         Advanced Data Augmentation         #
##############################################
def strong_aug(p=0.5):
    return A.Compose([
        A.RandomRotate90(p=0.2),
        A.Transpose(p=0.2),
        A.HorizontalFlip(p=0.5),
        A.VerticalFlip(p=0.5),
        A.OneOf([A.GaussNoise()], p=0.2),
        A.ShiftScaleRotate(p=0.2),
        A.OneOf([
            A.CLAHE(clip_limit=2),
            A.Sharpen(),
            A.Emboss(),
            A.RandomBrightnessContrast(),
        ], p=0.2),
        A.HueSaturationValue(p=0.2),
        A.RandomCrop(width=200, height=200, p=0.3),
    ], p=p)

def augment(aug, image):
    return aug(image=image)["image"]

class Aug(object):
    def __call__(self, img):
        aug = strong_aug(p=0.9)
        img_np = np.array(img)
        augmented = augment(aug, img_np)
        return Image.fromarray(augmented)

def normalize_data():
    mean = [0.485, 0.456, 0.406]
    std = [0.229, 0.224, 0.225]
    return {
        "train": transforms.Compose([
            Aug(),
            transforms.Resize((224, 224)),
            transforms.RandomHorizontalFlip(),
            transforms.ToTensor(),
            transforms.Normalize(mean, std),
            transforms.RandomErasing(p=0.2)
        ]),
        "valid": transforms.Compose([
            transforms.Resize((224, 224)),
            transforms.ToTensor(),
            transforms.Normalize(mean, std)
        ]),
        "test": transforms.Compose([
            transforms.Resize((224, 224)),
            transforms.ToTensor(),
            transforms.Normalize(mean, std)
        ])
    }

##############################################
#         Dataset Loading & Splitting        #
##############################################
class ExtractedFramesDataset(Dataset):
    def __init__(self, root_dir, transform=None, split="train", random_seed=42):
        """
        Expects root_dir to have the following structure:
        
        root_dir/
            celebdf/
                fake/    <-- label 1
                real/    <-- label 0
            dfdc/
                fake/
                real/
            ffpp/
                fake/
                real/
            realfake/
                fake/
                real/
        """
        self.samples = []
        self.transform = transform

        all_samples = []
        for subfolder in os.listdir(root_dir):
            subfolder_path = os.path.join(root_dir, subfolder)
            if not os.path.isdir(subfolder_path):
                continue
            for label_folder in os.listdir(subfolder_path):
                label_path = os.path.join(subfolder_path, label_folder)
                if not os.path.isdir(label_path):
                    continue
                label = 1 if label_folder.lower() == "fake" else 0
                for file in os.listdir(label_path):
                    if file.lower().endswith(('.png', '.jpg', '.jpeg')):
                        img_path = os.path.join(label_path, file)
                        all_samples.append((img_path, label))
        random.seed(random_seed)
        random.shuffle(all_samples)
        self.samples = all_samples
        print(f"Loaded {len(self.samples)} samples for the '{split}' split.")

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        img_path, label = self.samples[idx]
        image = Image.open(img_path).convert("RGB")
        if self.transform:
            image = self.transform(image)
        return image, label

##############################################
#     Helper to unify feature shapes         #
##############################################
def reduce_features(feat):
    if feat.ndim == 4:
        feat = F.adaptive_avg_pool2d(feat, 1).view(feat.size(0), -1)
    elif feat.ndim == 3:
        feat = feat.mean(dim=1)
    return feat

##############################################
#     Hybrid Model Definition (Swin + MobileNet)
##############################################
class SwinMobileNetHybrid(nn.Module):
    def __init__(self, num_classes=1):
        super(SwinMobileNetHybrid, self).__init__()
        # Swin Transformer
        self.swin = timm.create_model('swin_small_patch4_window7_224', pretrained=True)
        self.swin.head = nn.Identity()
        # MobileNetV3
        self.mobilenet = timm.create_model('mobilenetv3_large_100', pretrained=True)
        self.mobilenet.classifier = nn.Identity()
        dummy = torch.randn(1, 3, 224, 224)
        with torch.no_grad():
            swin_feat = self.swin.forward_features(dummy)
            swin_feat = reduce_features(swin_feat)
            mobilenet_feat = self.mobilenet.forward_features(dummy)
            mobilenet_feat = F.adaptive_avg_pool2d(mobilenet_feat, 1).view(mobilenet_feat.size(0), -1)
            self.swin_dim = swin_feat.shape[1]
            self.mobilenet_dim = mobilenet_feat.shape[1]
        self.fusion = nn.Sequential(
            nn.Linear(self.swin_dim + self.mobilenet_dim, 512),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(512, 256),
            nn.BatchNorm1d(256),
            nn.ReLU()
        )
        self.classifier = nn.Sequential(
            nn.Linear(256, 128),
            nn.ReLU(),
            nn.Dropout(0.2),
            nn.Linear(128, num_classes)
        )

    def forward(self, x):
        swin_feat = self.swin.forward_features(x)
        swin_feat = reduce_features(swin_feat)
        mobilenet_feat = self.mobilenet.forward_features(x)
        mobilenet_feat = F.adaptive_avg_pool2d(mobilenet_feat, 1).view(mobilenet_feat.size(0), -1)
        combined = torch.cat([swin_feat, mobilenet_feat], dim=1)
        fused = self.fusion(combined)
        out = self.classifier(fused)
        return out

##############################################
#       Label smoothing BCE loss             #
##############################################
class SmoothBCEWithLogitsLoss(nn.Module):
    def __init__(self, smoothing=0.1, pos_weight=None):
        super(SmoothBCEWithLogitsLoss, self).__init__()
        self.smoothing = smoothing
        self.pos_weight = pos_weight

    def forward(self, logits, target):
        target = target * (1 - self.smoothing) + 0.5 * self.smoothing
        return nn.functional.binary_cross_entropy_with_logits(logits, target, pos_weight=self.pos_weight)

##############################################
#          Mixup Data Functions              #
##############################################
def mixup_data(x, y, alpha=0.4):
    if alpha > 0:
        lam = np.random.beta(alpha, alpha)
    else:
        lam = 1
    batch_size = x.size()[0]
    index = torch.randperm(batch_size).to(device)
    mixed_x = lam * x + (1 - lam) * x[index, :]
    y_a, y_b = y, y[index]
    return mixed_x, y_a, y_b, lam

def mixup_criterion(criterion, pred, y_a, y_b, lam):
    return lam * criterion(pred, y_a) + (1 - lam) * criterion(pred, y_b)

##############################################
#     Training & Validation Loops            #
##############################################
def optimize_threshold(probs, true_labels):
    best_thresh = 0.5
    best_acc = 0.0
    for t in np.arange(0.0, 1.0, 0.01):
        preds = (probs >= t).astype(int)
        acc = accuracy_score(true_labels, preds)
        if acc > best_acc:
            best_acc = acc
            best_thresh = t
    return best_thresh, best_acc

##############################################
#                Main Training               #
##############################################
def main():
    data_transforms = normalize_data()
    train_dir = "/kaggle/input/deepfake-dataset/deepfake_dataset/train"
    val_dir = "/kaggle/input/deepfake-dataset/deepfake_dataset/validation"
    test_dir = "/kaggle/input/deepfake-dataset/deepfake_dataset/test"
    
    for path in [train_dir, val_dir, test_dir]:
        if not os.path.exists(path):
            raise FileNotFoundError(f"Dataset directory not found: {path}")
    
    train_dataset = ExtractedFramesDataset(train_dir, transform=data_transforms["train"], split="train")
    val_dataset = ExtractedFramesDataset(val_dir, transform=data_transforms["valid"], split="val")
    test_dataset = ExtractedFramesDataset(test_dir, transform=data_transforms["test"], split="test")

    train_loader = DataLoader(train_dataset, batch_size=16, shuffle=True, num_workers=2)
    val_loader = DataLoader(val_dataset, batch_size=16, shuffle=False, num_workers=2)
    test_loader = DataLoader(test_dataset, batch_size=16, shuffle=False, num_workers=2)

    model = SwinMobileNetHybrid(num_classes=1).to(device)
    logging.info("Swin + MobileNet Hybrid Model created")

    optimizer = optim.AdamW(model.parameters(), lr=1e-5, weight_decay=1e-4)
    num_epochs = 15
    scheduler = optim.lr_scheduler.LambdaLR(optimizer, lr_lambda=lambda epoch: 0.95 ** epoch)

    pos_weight_value = 10667 / 9333
    criterion = SmoothBCEWithLogitsLoss(
        smoothing=0.1,
        pos_weight=torch.tensor([pos_weight_value]).to(device)
    )

    best_val_loss = float('inf')
    patience = 5
    trigger_times = 0

    train_history = {"loss": [], "acc": []}
    val_history = {"loss": [], "acc": []}

    for epoch in range(num_epochs):
        model.train()
        running_loss = 0.0
        train_correct = 0
        train_total = 0
        train_bar = tqdm(train_loader, desc=f"Epoch {epoch+1}/{num_epochs} [Train]", leave=False)
        for images, labels in train_bar:
            images = images.to(device)
            labels = labels.float().unsqueeze(1).to(device)
            optimizer.zero_grad()

            if random.random() < 0.5:
                images, targets_a, targets_b, lam = mixup_data(images, labels)
                outputs = model(images)
                loss = mixup_criterion(criterion, outputs, targets_a, targets_b, lam)
            else:
                outputs = model(images)
                loss = criterion(outputs, labels)

            loss.backward()
            optimizer.step()

            running_loss += loss.item() * images.size(0)
            preds = (torch.sigmoid(outputs) > 0.5).int()
            train_correct += (preds.cpu() == labels.cpu().int()).sum().item()
            train_total += labels.size(0)
            train_bar.set_postfix(loss=f"{loss.item():.4f}", acc=f"{(train_correct/train_total):.4f}")

        epoch_loss = running_loss / train_total
        epoch_acc = train_correct / train_total
        train_history["loss"].append(epoch_loss)
        train_history["acc"].append(epoch_acc)

        # Validation
        model.eval()
        val_loss = 0.0
        val_correct = 0
        val_total = 0
        val_probs = []
        val_true = []
        val_bar = tqdm(val_loader, desc=f"Epoch {epoch+1}/{num_epochs} [Val]", leave=False)
        with torch.no_grad():
            for images, labels in val_bar:
                images = images.to(device)
                labels = labels.float().unsqueeze(1).to(device)
                outputs = model(images)
                loss = criterion(outputs, labels)
                val_loss += loss.item() * images.size(0)
                probs = torch.sigmoid(outputs)
                val_probs.extend(probs.cpu().numpy().flatten())
                preds = (probs > 0.5).int()
                val_correct += (preds.cpu() == labels.cpu().int()).sum().item()
                val_total += labels.size(0)
                val_true.extend(labels.cpu().numpy().flatten())

        epoch_val_loss = val_loss / val_total
        epoch_val_acc = val_correct / val_total
        val_history["loss"].append(epoch_val_loss)
        val_history["acc"].append(epoch_val_acc)
        scheduler.step()

        best_thresh, _ = optimize_threshold(np.array(val_probs), np.array(val_true))
        print(f"\nEpoch [{epoch+1}/{num_epochs}]")
        print(f"Train Loss: {epoch_loss:.4f}, Train Acc: {epoch_acc:.4f}")
        print(f"Val   Loss: {epoch_val_loss:.4f}, Val Acc: {epoch_val_acc:.4f}, Best Thresh: {best_thresh:.2f}")

        if epoch_val_loss < best_val_loss:
            best_val_loss = epoch_val_loss
            best_model_wts = copy.deepcopy(model.state_dict())
            trigger_times = 0
            torch.save(model.state_dict(), "best_swin_mobilenet.pth")
            logging.info("Best model saved")
        else:
            trigger_times += 1
            if trigger_times >= patience:
                logging.info("Early stopping!")
                break

        checkpoint = {
            "epoch": epoch,
            "model_state_dict": model.state_dict(),
            "optimizer_state_dict": optimizer.state_dict(),
            "scheduler_state_dict": scheduler.state_dict(),
            "best_val_loss": best_val_loss,
            "train_history": train_history,
            "val_history": val_history,
        }
        torch.save(checkpoint, "checkpoint_hybrid.pth")
        logging.info(f"Checkpoint saved at epoch {epoch+1}")

    def show_plot(metric_history, metric_name):
        plt.figure(figsize=(10,5))
        plt.plot(metric_history, label=f"{metric_name}")
        plt.title(f"{metric_name} Trend")
        plt.xlabel("Epoch")
        plt.ylabel(metric_name)
        plt.legend()
        plt.show()

    show_plot(train_history["loss"], "Train Loss")
    show_plot(train_history["acc"], "Train Accuracy")
    show_plot(val_history["loss"], "Validation Loss")
    show_plot(val_history["acc"], "Validation Accuracy")

    model.load_state_dict(torch.load("best_swin_mobilenet.pth"))
    model.eval()
    test_probs = []
    all_preds = []
    all_labels = []
    test_loss = 0.0
    test_correct = 0
    test_total = 0

    with torch.no_grad():
        for images, labels in tqdm(test_loader, desc="Testing", leave=False):
            images = images.to(device)
            labels = labels.float().unsqueeze(1).to(device)
            outputs = model(images)
            loss = criterion(outputs, labels)
            test_loss += loss.item() * images.size(0)
            probs = torch.sigmoid(outputs)
            test_probs.extend(probs.cpu().numpy().flatten())
            preds = (probs > best_thresh).int()
            all_preds.extend(preds.cpu().numpy().flatten())
            all_labels.extend(labels.cpu().numpy().flatten())
            test_correct += (preds.cpu() == labels.cpu().int()).sum().item()
            test_total += labels.size(0)

    test_loss /= len(test_dataset)
    accuracy_metric = accuracy_score(np.array(all_labels).astype(int), np.array(all_preds))
    precision_metric = precision_score(np.array(all_labels).astype(int), np.array(all_preds))
    recall_metric = recall_score(np.array(all_labels).astype(int), np.array(all_preds))
    f1_metric = f1_score(np.array(all_labels).astype(int), np.array(all_preds))
    conf_matrix = confusion_matrix(np.array(all_labels).astype(int), np.array(all_preds))
    try:
        auc_metric = roc_auc_score(np.array(all_labels).astype(int), test_probs)
    except:
        auc_metric = "N/A"
    fpr, tpr, thresholds = roc_curve(np.array(all_labels).astype(int), test_probs)
    try:
        eer_metric = brentq(lambda x: 1. - x - interp1d(fpr, tpr)(x), 0., 1.)
    except:
        eer_metric = 1.0

    print("\n=== Test Evaluation Metrics ===")
    print(f"Test Loss: {test_loss:.4f}")
    print(f"Test Accuracy: {accuracy_metric:.4f}")
    print(f"Precision: {precision_metric:.4f}")
    print(f"Recall: {recall_metric:.4f}")
    print(f"F1 Score: {f1_metric:.4f}")
    print(f"ROC-AUC: {auc_metric}")
    print(f"EER: {eer_metric:.4f}")
    print("Confusion Matrix:")
    print(conf_matrix)
    print("Classification Report:")
    print(classification_report(np.array(all_labels).astype(int), np.array(all_preds), digits=4))


if __name__ == "__main__":
    main()
```
